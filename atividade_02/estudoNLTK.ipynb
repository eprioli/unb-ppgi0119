### Introdução

Quando trabalhamos com processamento de linguagem natural (NLP), a tokenização é um passo crucial. Ela envolve a divisão de um texto em unidades menores, como sentenças ou palavras, para facilitar a análise e o processamento. O NLTK, uma biblioteca poderosa para NLP em Python, oferece métodos como `sent_tokenize` e `word_tokenize` para realizar essa tarefa de maneira eficiente.

A seguir, uma breve explicação dessas funções:

from nltk.tokenize import word_tokenize, sent_tokenize

### Explicação

- **sent_tokenize**: Esta função é usada para dividir um texto em sentenças individuais. Se você tiver um parágrafo, `sent_tokenize` pode identificar e separar cada sentença, levando em consideração pontuações e regras gramaticais.
- **word_tokenize**: Esta função é utilizada para dividir sentenças em palavras. Ela reconhece pontuações, abreviações e outros elementos, garantindo uma tokenização precisa das palavras no texto.

## Segue um código básico mostrando como usar a biblioteca do NLTK

```python

import nltk
print("Versão do nltk:",nltk.__version__)
print("NLTK está funcionando!")

```

```python
# Adiciona o diretório onde os dados do NLTK foram baixados
nltk.data.path.append('/home/eprioli/nltk_data')

from nltk.tokenize import word_tokenize, sent_tokenize

texto = "Olá! Como você está? Este é um exemplo de tokenização."
print("Texto completo:", texto)

# Dividir o texto em sentenças
sentencas = sent_tokenize(texto)
print("Texto em sentenças:", sentencas)

# Dividir o texto em palavras
palavras = word_tokenize(texto)
print("Texto em palavras:", palavras)
```

```python
import nltk
from nltk.corpus import stopwords

# Baixe a lista de stop words (apenas na primeira vez)
nltk.download('stopwords')

# Obtenha a lista de stop words para o idioma inglês
stop_words = set(stopwords.words('english'))

texto = "This is an example sentence showing off stop word filtration."
palavras = nltk.word_tokenize(texto)

# Filtrando as stop words
palavras_filtradas = [palavra for palavra in palavras if palavra.lower() not in stop_words]

print("Texto original:", palavras)
print("Texto sem stop words:", palavras_filtradas)
```

### O stemming é uma técnica de NLP que reduz as palavras a suas raízes ou radicais, removendo sufixos e prefixos.

Exemplo de Stemming com o NLTK

Primeiro, certifique-se de que o NLTK está instalado e importado:

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
# Inicializando o stemmer
stemmer = PorterStemmer()

# Texto de exemplo
texto = "Playing played player plays plays. We are enjoying our studies."

# Tokenizando as palavras
palavras = word_tokenize(texto)

# Aplicando o stemming em cada palavra
palavras_stemmed = [stemmer.stem(palavra) for palavra in palavras]

# Exibindo o resultado
print("Palavras originais:", palavras)
print("Palavras após o stemming:", palavras_stemmed)
```
